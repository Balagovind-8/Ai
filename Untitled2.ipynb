```python
import torch.nn as nn
import torch.optim as optim

# Define the model class
class QuadraticModel(nn.Module):
    def __init__(self):
        super().__init__() # This line should have consistent indentation with those below
        self.a = nn.Parameter(torch.randn(()))
        self.b = nn.Parameter(torch.randn(()))
        self.c = nn.Parameter(torch.randn(()))

    def forward(self, x):
        # ReLU applied to model output
        return torch.relu(self.a * x**2 + self.b * x + self.c)

# Initialize the model
model = QuadraticModel()

# Define loss function and optimizer
loss_fn = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=1e-4)

# Training loop
for epoch in range(10000):
    y_pred = model(x)              # forward pass
    loss = loss_fn(y_pred, y)      # compute loss

    optimizer.zero_grad()          # reset gradients
    loss.backward()                # backpropagation
    optimizer.step()               # update parameters

    if epoch % 1000 == 0:
        print(f"Epoch {epoch}, Loss: {loss.item()}")

        # Print learned coefficients
        print(f"\nLearned a: {model.a.item():.4f}, b: {model.b.item():.4f}, c: {model.c.item():.4f}")
```
